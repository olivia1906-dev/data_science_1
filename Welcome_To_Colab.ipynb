```
#reverse of a number
j = int(input("Enter a number:"))

original_num = j
reversed_num = 0
if j < 0:
    sign = -1
    j = abs(j)
else:
    sign = 1

while j > 0:
    remainder = j % 10
    reversed_num = (reversed_num * 10) + remainder
    j = j // 10

print("Original number:", original_num)
print("Reversed number:", reversed_num * sign)
```

```
#to count digit in a number
num = int(input("Enter a number: "))
count = 0
if num == 0:
    count = 1
else:
    temp_num = abs(num)
    while temp_num > 0:
        temp_num //= 10
        count += 1

print(f"The number of digits in {num} is: {count}")
```
