```
#find longest word in a sentence
str2  = input("Enter any sentence:")
w = str2.split()
l = max(w , key = len)
print(l)
```

```
#find all longest words in a sentence
str2  = input("Enter any sentence:")
w = str2.split()
if w:
    max_len = len(max(w , key = len))
    longest_words = [word for word in w if len(word) == max_len]
    print(longest_words)
else:
    print("No words entered.")
```
