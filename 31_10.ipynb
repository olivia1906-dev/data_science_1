Read a Large CSV in Chunks & Compute Groupby Mean Across Chunks

```python
np.random.seed(42)
categories = ["A", "B", "C", "D","E"]
data = {
    "Category": np.random.choice(categories, size=10000),
    "Score": np.random.uniform(10, 100, size=10000)
}
pd.DataFrame(data).to_csv("large_data.csv", index=False)

chunk_size = 2000
totals = pd.Series(dtype=float)
counts = pd.Series(dtype=int)

for chunk in pd.read_csv("large_data.csv", chunksize=chunk_size):
    grouped = chunk.groupby("Category")["Score"]
    totals = totals.add(grouped.sum(), fill_value=0)
    counts = counts.add(grouped.count(), fill_value=0)
overall_mean = totals / counts

print("Overall Mean Per Category across all chunks:")
print(overall_mean)
```

Export to JSON (orient="records"), Read Back, and Check Equality

```python
df_original = pd.DataFrame({
    "A": [1, 2, 3],
    "B": ["apple", "banana", "cherry"]
})
json_str = df_original.to_json(orient="records")
print("JSON Output:\n", json_str)

df_reloaded = pd.read_json(json_str, orient="records")

is_equal = df_original.equals(df_reloaded)
print("\nAre the DataFrames equal using .equals()?", is_equal)
```
